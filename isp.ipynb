```python
import pandas as pd
import numpy as np
```

```python
dados = "https://www.ispdados.rj.gov.br/Arquivos/BaseDPEvolucaoMensalCisp.csv"
df = pd.read_csv(dados, encoding='latin1', sep=';')
df
```

```python
df.dtypes
```

```python
df.info()
```

```python
df.describe()
```

```python
# Agrupando por CISP e somando os furtos de celular
df_cisp = df.groupby('cisp')['furto_celular'].sum().reset_index()
# Ordenando do maior para o menor
df_cisp = df_cisp.sort_values(by='furto_celular', ascending=False)
df_cisp
```

```python
# Agrupando por CISP e somando os furtos de celular
df_cisp_roubo = df.groupby('cisp')['roubo_veiculo'].sum().reset_index()
# Ordenando do maior para o menor
df_cisp_roubo = df_cisp_roubo.sort_values(by='roubo_veiculo', ascending=False)
df_cisp_roubo.head(15)
```

```python
df_cisp_roubo.tail(15)
```

```python
# Usando .loc
df_capital = df.loc[df['munic'] == 'Rio de Janeiro'] 
df

```

```python
# Usando .query() — sintaxe alternativa mais legível
df_capital = df.query("munic == 'Rio de Janeiro'")
```

```python
# Repetindo o groupby apenas para a capital
df_cisp_capital = df_capital.groupby('cisp')['furto_celular'].sum().reset_index()
df_cisp_capital = df_cisp_capital.sort_values(by='furto_celular', ascending=False)
```

```python
df_cisp_furto = df_capital.groupby('cisp')['furto_veiculos'].sum().reset_index()
df_cisp_furto = df_cisp_furto.sort_values(by='furto_veiculos', ascending=False)
df_cisp_furto.head(10)
```

```python
array_cisp_capital_furto = np.array(df_cisp_furto['furto_veiculos'])
array_cisp_capital_furto 
```

```python
media = np.mean(array_cisp_capital_furto)
mediana = np.median(array_cisp_capital_furto)
q1 = np.percentile(array_cisp_capital_furto, 25)
q2 = np.percentile(array_cisp_capital_furto, 50)
q3 = np.percentile(array_cisp_capital_furto, 75)
print('Q1:', q1)
print('Q2:', q2)
print('Q3:', q3)

print(f'Média {round(media,2)}\nMediana {mediana}')
```

```python
top_25 = df_cisp_furto.loc[df_cisp_furto['furto_veiculos'] > q3]
top_25
```
